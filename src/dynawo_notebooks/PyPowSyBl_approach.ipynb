```python
# Cell 1: Environment Setup
import os
import pypowsybl as pp
from dynawo_notebooks.Scripts.core.mo_topology import MoTopologyToolkit
from IPython.display import SVG, display

# --- Configuration ---

# 1. Define tgetComponents({self.model_name}he directory containing your models
# Assuming your models are in a subfolder named "Models" relative to this notebook
MODELS_DIR = os.path.abspath("Models")

# 2. Define the main model name
MODEL_NAME = "Dynawo.Examples.BESS.WECC.MyBESS_static"

# 3. Path to the Dynawo package.mo
# (Based on your previous configuration, verify this path exists)
DYNAWO_PKG_PATH = "/home/guiu/Projects/Dynawo/nightly/dynawo/ddb/Dynawo/package.mo"

# 4. List of files to load (Order matters: Dependencies -> Main Model)
# We list the file previously referenced. If this model depends on others
# (like BESS_init.mo), add them to this list BEFORE "MyBESS_static.mo".
LOCAL_FILES = ["MyBESS_static.mo"]

# --- Initialization ---

# Initialize our expert parser with the new manual loading sequence
toolkit = MoTopologyToolkit(
    source_dir=MODELS_DIR,
    model_name=MODEL_NAME,
    dynawo_pkg_path=DYNAWO_PKG_PATH,
    local_files_list=LOCAL_FILES,
)

print(f"Toolkit initialized successfully for model: {MODEL_NAME}")
```

```python
# Cell 2: Extraction and Export
# Get structured electrical data in SI units
parsed_data = toolkit.parse_electrical_data()

# Save to the easy-to-manipulate standard format (JSON)
toolkit.export_to_standard_json(parsed_data, "grid_export.json")

print("Topology extracted and saved to grid_export.json successfully.")
```

```python
# Cell 3: Build PowSybl Network
network = toolkit.build_powsybl_network(parsed_data)

# Print a summary of the created network
print(f"Network created with {len(network.get_buses())} buses.")
print(f"Voltage Levels: {list(network.get_voltage_levels().index)}")
```

```python
# Cell 4: Visualization
# Generate a Network Area Diagram (NAD)
# This leverages the pypowsybl-diagram library internally
nad_svg = network.get_network_area_diagram()

print("Displaying the generated Network Topology Graph:")
display(SVG(str(nad_svg)))
```
